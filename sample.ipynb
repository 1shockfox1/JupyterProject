Задача 1

```python
from random import randint



Kolvo=int(input("Enter Kolvo number: "))
mas = tuple(randint(10, 20)
            for i in range(Kolvo))

print("Массив:", mas)

found=False
for i in range(Kolvo):
    if mas[i]==1 and mas[i+1]==1:
        found=True
        print("Два нуля")
        break

if not found:
    print("в массиве нет нулей подряд")
```

Задача 2

```python
n,m=5,5

mas2=tuple(tuple([None]*n for i in range(m)))
for i in range(n):
    for j in range(m):
        if (i + j) % 2 == 0:
            mas2[i][j] = "."
        else:
            mas2[i][j] = "*"
for i in range(n):
    for j in range(m):
        print(mas2[i][j], end=" | ")
    print()
```

Задача 4

```python
only_p = set(range(11))

only_c = set(range(11, 15))

only_s = set(range(15, 17))

only_pc = set(range(17, 22))

only_ps = set(range(22, 25))

only_cs = set(range(25, 26))


P = only_p | only_pc | only_ps
C = only_c | only_pc | only_cs
S = only_s | only_pc | only_cs


visit = P | C | S

sick = 3

total = sick + len(visit)

print("Посетили хотя бы одно место: ", len(visit))
print("VSEGO: ", total)
```
