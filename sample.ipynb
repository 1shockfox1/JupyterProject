Задача 1

```python
from random import randint



Kolvo=int(input("Enter Kolvo number: "))
mas = tuple(randint(10, 20)
            for i in range(Kolvo))

print("Массив:", mas)

found=False
for i in range(Kolvo):
    if mas[i]==1 and mas[i+1]==1:
        found=True
        print("Два нуля")
        break

if not found:
    print("в массиве нет нулей подряд")
```

Задача 2

```python
n,m=5,5

mas2=tuple(tuple([None]*n for i in range(m)))
for i in range(n):
    for j in range(m):
        if (i + j) % 2 == 0:
            mas2[i][j] = "."
        else:
            mas2[i][j] = "*"
for i in range(n):
    for j in range(m):
        print(mas2[i][j], end=" | ")
    print()
```
